## 3. NumPy Array Analysis

NumPy arrays allow us to perform fast numerical calculations and statistical operations on student performance data.

```python
math_scores = np.array(df['Math_Score'])
science_scores = np.array(df['Science_Score'])
english_scores = np.array(df['English_Score'])
computer_scores = np.array(df['Computer_Score'])
percentages = np.array(df['Percentage'])
study_hours = np.array(df['Study_Hours'])
attendance = np.array(df['Attendance'])
```

```python
math_scores[:10]
```

## 4. Descriptive Statistics

We will calculate the main statistical measures for student performance using NumPy.

```python
#Mean
```

```python
mean_math = np.mean(math_scores)
mean_science = np.mean(science_scores)
mean_english = np.mean(english_scores)
mean_computer = np.mean(computer_scores)

print("Average Math Score:", round(mean_math, 2))
print("Average Science Score:", round(mean_science, 2))
print("Average English Score:", round(mean_english, 2))
print("Average Computer Score:", round(mean_computer, 2))
```

```python
#Median
```

```python
print("Math Median:", np.median(math_scores))
print("Science Median:", np.median(science_scores))
print("English Median:", np.median(english_scores))
print("Computer Median:", np.median(computer_scores))
```

```python
#Standard Deviation
```

```python
print("Math Standard Deviation:", round(np.std(math_scores), 2))
print("Science Standard Deviation:", round(np.std(science_scores), 2))
print("English Standard Deviation:", round(np.std(english_scores), 2))
print("Computer Standard Deviation:", round(np.std(computer_scores), 2))
```

```python
#Variance
```

```python
print("Math Variance:", round(np.var(math_scores), 2))
print("Science Variance:", round(np.var(science_scores), 2))
print("English Variance:", round(np.var(english_scores), 2))
print("Computer Variance:", round(np.var(computer_scores), 2))
```

```python
#Minimum & Maximum
```

```python
print("Highest Math Score:", np.max(math_scores))
print("Lowest Math Score:", np.min(math_scores))

print("Highest Science Score:", np.max(science_scores))
print("Lowest Science Score:", np.min(science_scores))

print("Highest English Score:", np.max(english_scores))
print("Lowest English Score:", np.min(english_scores))

print("Highest Computer Score:", np.max(computer_scores))
print("Lowest Computer Score:", np.min(computer_scores))
```

```python
#Percentiles
```

```python
print("25th Percentile:", np.percentile(percentages, 25))
print("50th Percentile:", np.percentile(percentages, 50))
print("75th Percentile:", np.percentile(percentages, 75))
print("90th Percentile:", np.percentile(percentages, 90))
```

## 5. Subject-wise Performance Analysis

In this section, we compare student performance across different subjects and identify the strongest and weakest subjects.

```python
#Calculate average score for each subject
```

```python
subject_means = {
    "Math": np.mean(math_scores),
    "Science": np.mean(science_scores),
    "English": np.mean(english_scores),
    "Computer": np.mean(computer_scores)
}

subject_means
```

```python
#Find the highest-performing subject
```

```python
best_subject = max(subject_means, key=subject_means.get)

print("Highest Average Score Subject:", best_subject)
print("Average Score:", round(subject_means[best_subject], 2))
```

```python
#Find the lowest-performing subject
```

```python
weakest_subject = min(subject_means, key=subject_means.get)

print("Lowest Average Score Subject:", weakest_subject)
print("Average Score:", round(subject_means[weakest_subject], 2))
```

```python
#Subject-wise Statistics Table
```

```python
subject_statistics = pd.DataFrame({
    "Subject": ["Math", "Science", "English", "Computer"],
    "Mean": [
        np.mean(math_scores),
        np.mean(science_scores),
        np.mean(english_scores),
        np.mean(computer_scores)
    ],
    "Median": [
        np.median(math_scores),
        np.median(science_scores),
        np.median(english_scores),
        np.median(computer_scores)
    ],
    "Standard_Deviation": [
        np.std(math_scores),
        np.std(science_scores),
        np.std(english_scores),
        np.std(computer_scores)
    ],
    "Minimum": [
        np.min(math_scores),
        np.min(science_scores),
        np.min(english_scores),
        np.min(computer_scores)
    ],
    "Maximum": [
        np.max(math_scores),
        np.max(science_scores),
        np.max(english_scores),
        np.max(computer_scores)
    ]
})

subject_statistics.round(2)
```

## 6. Grade Distribution

We will analyze how many students belong to each grade category.

```python
#Count students in each grade
```

```python
grade_counts = df["Grade"].value_counts().sort_index()

grade_counts
```

```python
#Calculate percentage of students in each grade
```

```python
grade_percentage = (grade_counts / len(df)) * 100

grade_percentage.round(2)
```

```python
#Find the most common grade
```

```python
most_common_grade = df["Grade"].mode()[0]

print("Most Common Grade:", most_common_grade)
print("Number of Students:", grade_counts[most_common_grade])
```

## 7. Top 10 Students

Identify the 10 students with the highest overall percentage.

```python
top_10_students = df.sort_values(
    by="Percentage",
    ascending=False
).head(10)

top_10_students[
    [
        "Student_ID",
        "Gender",
        "Study_Hours",
        "Attendance",
        "Percentage",
        "Grade"
    ]
]
```

```python
#Bottom 10 Students
```

```python
bottom_10_students = df.sort_values(
    by="Percentage",
    ascending=True
).head(10)

bottom_10_students[
    [
        "Student_ID",
        "Gender",
        "Study_Hours",
        "Attendance",
        "Percentage",
        "Grade"
    ]
]
```

```python
#Students Above Average
```

```python
average_percentage = np.mean(percentages)

above_average = percentages > average_percentage

print("Average Percentage:", round(average_percentage, 2))
print("Students Above Average:", np.sum(above_average))
```

```python
above_average_percentage = (np.sum(above_average) / len(df)) * 100

print(
    "Percentage of Students Above Average:",
    round(above_average_percentage, 2),
    "%"
)
```

```python
#Students Below Average
```

```python
below_average = percentages < average_percentage

print("Students Below Average:", np.sum(below_average))
```

## 8. Study Hours vs Academic Performance

We will analyze whether students who spend more time studying tend to achieve higher percentages.

```python
#Calculate correlation
```

```python
study_percentage_corr = np.corrcoef(study_hours, percentages)[0, 1]

print("Correlation between Study Hours and Percentage:",
      round(study_percentage_corr, 3))
```

The above data is positive relationship

## 9. Attendance vs Academic Performance

We will analyze whether students with higher attendance tend to achieve better academic results.

```python
attendance_percentage_corr = np.corrcoef(
    attendance,
    percentages
)[0, 1]

print("Correlation between Attendance and Percentage:",
      round(attendance_percentage_corr, 3))
```

```python
#Average Performance by Study-Hour Group
```

```python
df["Study_Hour_Group"] = pd.cut(
    df["Study_Hours"],
    bins=[0, 2, 4, 6, 8, 10],
    labels=["0-2 Hours", "2-4 Hours", "4-6 Hours", "6-8 Hours", "8-10 Hours"]
)
```

```python
study_group_analysis = df.groupby(
    "Study_Hour_Group",
    observed=False
)["Percentage"].mean()

study_group_analysis.round(2)
```

```python
#Average Performance by Attendance Group
```

```python
df["Attendance_Group"] = pd.cut(
    df["Attendance"],
    bins=[0, 60, 70, 80, 90, 100],
    labels=[
        "Below 60%",
        "60-70%",
        "70-80%",
        "80-90%",
        "90-100%"
    ]
)
```

```python
attendance_group_analysis = df.groupby(
    "Attendance_Group",
    observed=False
)["Percentage"].mean()

attendance_group_analysis.round(2)
```

## 10. Performance Classification Using NumPy

Students will be classified into performance categories based on their percentage.

```python
df["Performance_Level"] = np.where(
    df["Percentage"] >= 75,
    "High Performer",
    np.where(
        df["Percentage"] >= 50,
        "Average Performer",
        "Low Performer"
    )
)
```

```python
df[
    [
        "Student_ID",
        "Percentage",
        "Performance_Level"
    ]
].head(10)
```

```python
#Count Performance Levels
```

```python
performance_counts = df["Performance_Level"].value_counts()

performance_counts
```

```python
performance_percentage = (
    performance_counts / len(df)
) * 100

performance_percentage.round(2)
```

## 11. Gender-wise Performance Analysis

We will compare the average academic performance of students across genders.

```python
gender_performance = df.groupby("Gender")["Percentage"].agg(
    ["mean", "median", "std", "min", "max"]
)

gender_performance.round(2)
```

```python
#Find the Best Performing Gender Group
```

```python
best_gender = gender_performance["mean"].idxmax()

print("Higher Average Performance:", best_gender)
print(
    "Average Percentage:",
    round(gender_performance.loc[best_gender, "mean"], 2)
)
```
